```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Basic syntax

# plt.pie(values)
# plt.show()
```

```python
df = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Revenue": [45000,52000,48000,61000,72000,68000,75000,82000,79000,88000,92000,97000],
    "Orders": [120,140,130,160,180,170,200,220,210,240,260,280],
    "Returns": [5,7,6,8,10,9,11,12,10,13,14,15],
    "Customers": [90,110,100,130,150,145,170,190,180,210,230,250],
    "Region": ["North","North","South","South","West","West","East","East","North","South","West","East"]
})
```

```python
# Group by Region
region_rev = df.groupby("Region")["Revenue"].sum()
```

```python
region_rev
```

1. Simple Pie Chart

```python
plt.figure(figsize=(6,6))
plt.pie(region_rev.values)
plt.title("region wise pie Chart")
plt.show
```

2. Adding Labels 

```python
plt.figure(figsize=(6,6))

plt.pie(region_rev.values, labels= region_rev.index)
plt.title("Region wise Revenue Share")
plt.show()
```

3. Percentage Show

```python

plt.figure(figsize=(6,6))
plt.pie(region_rev.values, labels= region_rev.index, autopct= "%1.1f%%") 
plt.title("Region Wise Revenue Share")
plt.show()

# meaning of autopct 
# %1.1f%%
# 1 = minimum digit
# .1 = decimal place
# f = float
# %% = percentage symbol
```

4. Slice Highlight

```python
explode = [0.1, 0, 0, 0] 
plt.pie(region_rev.values,
        labels=region_rev.index,
        autopct="%1.1f%%",
        explode=explode)

plt.title("Region Wise Revenue Share")
plt.show()
```

```python
explode = [0,0.2,0,0]
plt.pie(region_rev.values,labels= region_rev.index,explode=explode,autopct = "%1.2f%%")
plt.title("Region WIse Revenue Share")
plt.show()
```

5. Adding colors to slice

```python
colors = ["skyblue","purple","orange","pink"]
explode = [0,0.1,0,0]
plt.pie(region_rev.values, labels= region_rev.index, autopct= "%1.1f%%",colors=colors,explode=explode)
plt.title("Region wise Revenue Chart")
plt.show()
```

6. Professional Pie Chart (Complete)

```python
plt.figure(figsize=(6,7))
plt.pie(region_rev.values, 
        labels=region_rev.index, # labeling of the data 
        autopct="%1.1f%%", # to show percentage of data 
        startangle=90, # starting from 90 degree 
        explode=[0.02]*len(region_rev)) # explode with 0.02 with length of data 
plt.title("Region Wise Revenue Chart")
plt.axis("equal") # to make perfect circle 
plt.show()
```
