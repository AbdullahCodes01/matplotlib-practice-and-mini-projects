```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Revenue": [45000,52000,48000,61000,72000,68000,75000,82000,79000,88000,92000,97000],
    "Orders": [120,140,130,160,180,170,200,220,210,240,260,280],
    "Returns": [5,7,6,8,10,9,11,12,10,13,14,15],
    "Customers": [90,110,100,130,150,145,170,190,180,210,230,250],
    "Region": ["North","North","South","South","West","West","East","East","North","South","West","East"]
})
```

```python
df
```

```python
plt.bar(df["Month"], df["Revenue"]) #Basic Syntax
plt.show()
```

```python
# Month vs Orders Bar Chart
plt.figure(figsize=(10,5))
plt.bar(df["Month"],df["Revenue"])
plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.show()
```

## Bar Chart Styling

```python
plt.figure(figsize=(10,5))
plt.bar(df["Month"],df["Revenue"],color = 'skyblue',edgecolor = "black")
plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.grid(axis="y")
plt.show()
```

```python
# Revenue Bar chart
plt.figure(figsize=(10,6))
plt.bar(df["Month"],df["Revenue"],color = 'purple',edgecolor = "black")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(11,6))
plt.barh(df["Month"],df["Returns"],color ='lightsalmon',edgecolor ="saddlebrown")
plt.title("Monthly Returns")
plt.xlabel("Revenue")
plt.ylabel("Month")
plt.grid(axis="x")
plt.show()
```

## Region Wise Revenue

```python
region_rev = df.groupby("Region")["Revenue"].sum()
```

```python
region_rev
```

```python
plt.figure(figsize=(6,4))
plt.bar(region_rev.index, region_rev.values,color ="teal",edgecolor ="black",width=0.5)
plt.title("Region Wise Revenue")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(len(df["Month"]))

plt.figure(figsize=(10,5))

plt.bar(x - 0.2, df["Revenue"], width=0.4, label="Revenue")
plt.bar(x + 0.2, df["Orders"], width=0.4, label="Orders")

plt.xticks(x,df["Month"].tolist())

plt.title("Revenue vs Orders Comparison")
plt.xlabel("Month")
plt.ylabel("Value")
plt.legend()
plt.grid(axis="y")

plt.show()
```
