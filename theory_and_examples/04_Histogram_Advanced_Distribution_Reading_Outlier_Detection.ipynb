```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.DataFrame({
"salary": [30000,35000,40000,45000,50000,60000,70000,80000,900000],

})
```

Code — Distribution Reading Practice

```python
plt.hist(df["salary"],bins= 8)
plt.title("Salary Distribution")
plt.show()
```

```python
# Skewness Quick Logic

# Right Skew Means
# Mean > Median

# Left Skew Means
# Mean < Median

# Pandas Quick Check
print(df["salary"].mean())
print(df["salary"].median())
```
