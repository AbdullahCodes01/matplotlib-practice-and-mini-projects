```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

viz_df = pd.DataFrame({
    "Month": ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
    "Revenue": [45000,52000,48000,61000,72000,68000,75000,82000,79000,88000,92000,97000],
    "Orders": [120,140,130,160,180,170,200,220,210,240,260,280],
    "Returns": [5,7,6,8,10,9,11,12,10,13,14,15],
    "Customers": [90,110,100,130,150,145,170,190,180,210,230,250],
    "Region": ["North","North","South","South","West","West","East","East","North","South","West","East"]
})

viz_df
```

## LEVEL 1 — BASIC MATPLOTLIB

```python
# Q1 — Revenue Line Plot

# Plot Revenue vs Month
# Add:
# Title
# Labels
# Grid

plt.plot(viz_df["Month"],viz_df["Revenue"])
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid()
plt.show()
```

```python
# Q2 — Styling Practice

# Revenue plot with:
# Marker
# Custom line style
# Custom color
# Set figure size = (10,5)
plt.figure(figsize=(10,5))
plt.plot(viz_df["Month"],viz_df["Revenue"],marker = "o",linestyle = "--",color = "purple")
plt.title("Monthly Revenue",color = "red")
plt.xlabel("Month",color = "orange")
plt.ylabel("Revenue",color = "brown")
plt.grid()
plt.show()
```

## LEVEL 2 — MULTI LINE & COMPARISON

```python
# Q4 — Revenue vs Orders

# plot in same graph:
# Revenue
# Orders
# Add legend
plt.figure(figsize=(10,5))
plt.plot(viz_df["Month"],viz_df["Revenue"],marker = "o",linestyle = "--",color = "purple",label = "Revenue")
plt.plot(viz_df["Month"],viz_df["Orders"],marker = "o",linestyle = ":",color = "brown",label = "Orders")
plt.title("Monthly Revenue",color = "red")
plt.xlabel("Month",color = "orange")
plt.ylabel("Revenue",color = "brown")
plt.legend()
plt.grid()
plt.show()

# note:
#     Because Revenue and Orders are on different scales.
# I would use dual axis or normalization to compare trends properly after learning more .
```

```python
# Q5 — Highlight Highest Revenue Month
# highlight using Scatter highlight 
max_row = viz_df.loc[viz_df["Revenue"].idxmax()]

plt.figure(figsize=(10,5))

plt.plot(viz_df["Month"], viz_df["Revenue"],
         marker="o", linestyle="--")

plt.scatter(max_row["Month"], max_row["Revenue"], s=200)
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid()
plt.show()
```

```python
# Q6. Highlight Lowest Revenue Month
min_row = viz_df.loc[viz_df["Revenue"].idxmin()]

plt.figure(figsize=(10,5))
plt.plot(viz_df["Month"],viz_df["Revenue"],linestyle = ":",marker = "o",color ="brown")
plt.scatter(min_row["Month"],min_row["Revenue"],s=200)
plt.title("Lowest Revenue Month")
plt.xlabel("Month",color = "red")
plt.ylabel("Revenue",color ="purple")
plt.grid()
plt.show()

```

## LEVEL 3 — HISTOGRAM (Distribution Analysis)

```python
# Q7 — Revenue Histogram

# bins = 10
# grid add

plt.hist(viz_df["Revenue"],bins=10)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.show()
```

```python
# Orders Distribution Histogram

# bins = 8
# edgecolor add

plt.hist(viz_df["Orders"],bins=8,edgecolor = "black")
plt.title("Orders Distribution")
plt.xlabel("Orders")
plt.ylabel("Frequency")
plt.show()
```

```python
# Q9 — Returns Distribution

# Identify:
# Most common range

plt.figure(figsize=(8,5))

plt.hist(viz_df["Returns"], bins=6, edgecolor="black")

plt.title("Returns Distribution")
plt.xlabel("Returns")
plt.ylabel("Frequency")

plt.grid()
plt.show()
```

```python
# median > mode
# Median > Mean  →  Left Skew (Negative Skew)
# Mean   > Median → Right Skew (Positive Skew)
              # Slightly Left Skewed (Negative Skewed)

# The revenue distribution is slightly left-skewed, indicating that most months generated higher revenue, with a few lower revenue months pulling the mean downward.
```
